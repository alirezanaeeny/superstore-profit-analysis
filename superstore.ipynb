```
import pandas as pd 
import numpy as np
import seaborn as sns 
import matplotlib.pyplot as plt
import openpyxl
df=pd.read_csv("superstore.csv",encoding="cp1256")
df['Order Date']=pd.to_datetime(df['Order Date'])
df["Ship Date"]=pd.to_datetime(df["Ship Date"])
print(df.groupby('Discount')['Profit'].mean())
print(df.groupby("Category")["Profit"].mean())
print(df.groupby("Category")["Discount"].mean())
print(df.groupby("Segment")["Profit"].mean())
print(df.groupby("Segment")["Discount"].mean())
loss=df[(df["Category"]=="Furniture") & (df["Discount"]>0.2)]
print("تعداد سفارش هاي با تخفيف بالاي 20 درصد:",len(loss))
print("مجموع ضرر سفارش هاي بالاي 20 درصد تخفيف دردسته مبلمانن:",loss["Profit"].sum())
plt.figure(figsize=(7,7))
sns.lineplot(data=df, x='Discount', y='Profit',marker='o', color='darkred')
plt.axhline(0, color='gray', linestyle='--')
plt.title('Profit chart based on discount')
plt.xlabel('discount')
plt.ylabel('avg profit')
plt.show()
```
